### Data Changes:

- years_worked (int)
- sleep_hours (int)
- weekly_study_hours (int)
- household_size (int)
- pet_count (int)

```python
df['years_worked'].value_counts()
```

```python
df['years_worked'] = pd.to_numeric(
    df['years_worked'].astype('string').str.extract(r'(\d+(?:\.\d+)?)', expand=False),
    errors='coerce'
)
```

```python
df['sleep_hours'].value_counts()
```

```python
import re

def convert_range(value):
    """Convert a number or numeric range to a float; ranges become their average."""
    if pd.isna(value):
        return np.nan

    numbers = re.findall(r"\d+(?:\.\d+)?", str(value))
    if not numbers:
        return np.nan

    return float(np.mean([float(number) for number in numbers]))

df["sleep_hours"] = df["sleep_hours"].apply(convert_range)
```

```python
df['sleep_hours'].value_counts()
```

```python
# - weekly_study_hours (int)
# - household_size (int)
# - pet_count (int)

columns_to_fix = ['weekly_study_hours', 'household_size', 'pet_count']

for val in columns_to_fix:
    df[val] = df[val].apply(convert_range)
```
