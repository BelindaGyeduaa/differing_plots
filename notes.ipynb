### Scatter plots - used to visualize relationships between 2 or more variables

- `plt.scatter(x, y, marker, alpha, color, cmap, s=size)`

```python
plt.grid(alpha=0.3, linestyle='--')
plt.scatter(x, y, color='red', marker='*', s=20)
plt.show()
```

```python
x = np.array([5,7,8,7,2,17,2,9,4,11,12,9,6])
y = np.array([99,86,87,88,111,86,103,87,94,78,77,85,86])
sizes = np.array([20, 50, 100, 200, 500, 1000, 60, 90, 10, 300, 600, 800, 75])
```

```python
plt.figure(figsize=(6, 4))
plt.grid(alpha=0.3, linestyle='--')
plt.scatter(x, y, color='red', s=sizes, alpha=0.5)
plt.show()
```

```python
# Object Oriented (OO)
fig, axes = plt.subplots(figsize=(6, 4))


plt.grid(alpha=0.3, linestyle='--')
plt.scatter(x, y, color='red', s=sizes)
plt.show()
```

```python
x = np.random.randint(100, size=(100))
y = np.random.randint(100, size=(100))
colors = np.random.randint(100, size=(100))
sizes = np.random.randint(100, size=(100))

plt.scatter(x, y, c=colors, s=sizes, alpha=0.5, cmap='spring')
plt.title('chart title')
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.colorbar(label='Color value')
plt.gca().set_facecolor('lightblue')
plt.show()
```

```python
AgeGroup1 = range(40, 70) # ages 40 - 69
purchase_data = np.abs(np.random.rand(30) * 20)

plt.scatter(
    AgeGroup1,
    purchase_data,
    s=200,
    alpha=0.6,
    label='40-69 Years Old'
)
plt.scatter(AgeGroup1, purchase_data)

AgeGroup2 = range(20, 40) # ages 20 - 39
purchase_data2 = np.abs(np.random.rand(20) * 40)

plt.scatter(
    AgeGroup2,
    purchase_data2,
    s=200,
    alpha=0.6,
    label='20-39 Years Old'
)
plt.scatter(AgeGroup2, purchase_data2)
plt.title('Purchase by age groups')
plt.ylabel('Average purchase per weekend')
plt.xlabel('Age')
plt.legend()

plt.show()
```

### BarCharts - categorical comparisons

- bar chart
- histograms - quantitative
- hz barchars
- Stacked Barcharts
- Hz Stacked Barcharts
- Clustered Column/Barcharts
- Hz Clustered Column/Barcharts

```python
courses = ['Machine Learning', 'SQL Basics', 'Python Pro', ' Data Analytics', 'Cloud Databases']
num_of_students = [50, 90, 30, 130, 16]

plt.figure(figsize=(10, 8))
plt.barh(courses, num_of_students, color='purple', height=0.5)
plt.title('Number of Student Per Class')
plt.xlabel('Class Name')
plt.ylabel('Number of Students')

plt.show()
```

```python
plt.figure(figsize=(10, 8))
plt.bar(courses, num_of_students, color='green', width=0.5)
plt.title('Number of Student Per Class')
plt.xlabel('Class Name', loc='right')
plt.ylabel('Number of Students')

plt.show()
```

```python
courses = ['Machine Learning', 'SQL Basics', 'Python Pro', 'Data Analytics', 'Cloud Databases']
course_group1 = [50, 90, 30, 130, 16]
course_group2 = [25, 60, 180, 12, 50]

plt.figure(figsize=(10, 8))

plt.bar(courses, course_group1, width=0.3, label='Group 1')

plt.bar(courses, course_group2, width=0.3, bottom=course_group1, label='Group 2')

plt.title('Number of Student Per Class')
plt.xlabel('Class Name')
plt.ylabel('Number of Students')
plt.legend()

plt.show()
```

```python
years = ['2016', '2017', '2018', '2019']
rice_price = [57, 50, 47, 30]
wheat_price = [43, 50, 53, 70]
```

```python
plt.figure(figsize=(8, 6))

plt.bar(years, rice_price, width=0.3, label='Rice')

plt.bar(years, wheat_price, width=0.3, bottom=rice_price, label='Wheat')

plt.title('The Preference of Rice vs Wheat')
plt.xlabel('Years')
plt.ylabel('Preference')
plt.legend()

plt.show()
```

```python
years = ['2016', '2017', '2018', '2019']
rice =  np.array([57, 50, 47, 30])
wheat =  np.array([43, 50, 53, 70])
mustard =  np.array([10, 16, 13, 15])

plt.figure(figsize=(8, 6))

plt.bar(years, rice, width=0.3, label='Rice')

plt.bar(years, wheat, bottom=rice, width=0.3, label='Wheat')
plt.bar(years, mustard, bottom=rice+wheat, width=0.3, label='Mustard')

plt.title('The Preference of Rice vs Wheat vs Mustard')
plt.xlabel('Years')
plt.ylabel('Preference')
plt.legend()

plt.show()
```

```python
courses = ['Machine Learning', 'SQL Basics', 'Python Pro', 'Data Analytics', 'Cloud Databases']
course_group1 = [50, 90, 30, 130, 16]
course_group2 = [25, 60, 180, 12, 50]
width = 0.3  # predeclare the width of a column

# calculate column placement. Creating a list/array with the index values for each category
x = np.arange(len(courses))

plt.grid(alpha=0.5)
plt.bar(x - width/2, course_group1, width=width, label='Group 1')
plt.bar(x + width/2, course_group2, width=width, label='Group 2')

plt.xticks(x, courses, rotation=45)
plt.legend()

plt.show()
```

```python
# Sample data
data = {
    'CoursesName': ['C Programming Course', 'C++ Programming Course', 'Java Programming Course', 'Python Programming Course'],
    'Student Group 1': [90, 60, 130, 600],
    'Student Group 2': [100, 30, 160, 690]
}

df = pd.DataFrame(data)

width = 0.35

index = df.index

plt.bar(index - width/2, df['Student Group 1'], width=width, label='Student Group 1')
plt.bar(index + width/2, df['Student Group 2'], width=width, label='Student Group 2')
plt.xticks(index, df['CoursesName'], rotation=45, ha='right')
plt.legend()

plt.show()
```
